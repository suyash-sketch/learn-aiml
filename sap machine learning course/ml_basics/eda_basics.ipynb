```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([10,20,30,40,50])
```

```python
plt.plot(x,y,c='g')
```

Line Plot

```python
x = np.array([1,2,3,4,5])
y = x*x*x
```

```python
plt.plot(x,y, 'bo', linestyle = 'dashed', linewidth = 5, markersize = 10)
```

```python
x = np.arange(0,4*np.pi, 0.001)
y = np.sin(x)

plt.plot(x,y)
```

```python
x = np.array([55,62,59,45,70])
y = np.array([100,105,180,150,160])

plt.plot(x,y)
```

Bar Plot

```python
height = [3,22,5,18,45]
bar = ['A','B','C','D','E']
x_pos = np.arange(len(bar))
plt.bar(x_pos, height, color = ('black','red','green','blue','cyan'))
```

```python
x = np.random.normal(100,0.6,250) #mean,std deviation, size of array
plt.hist(x)
plt.show()
```

```python
x = np.random.random(10)

y = np.random.randint(low=10,high=30, size=10)
plt.scatter(x,y)
plt.show()
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df['who'].value_counts()
```

```python
sns.countplot(data=df, x='who')
```

```python
sns.countplot(data=df, x='class')
plt.plot()
```

```python
df['alive'].value_counts().plot(kind='pie', autopct = '%1.1f%%')
plt.title('Pie chart of alive')
plt.show()
```

```python
sns.countplot(data=df, x='sex', hue='who')
```

```python
sns.countplot(data=df,x='embarked', hue='alive')
```

```python
alive_class = pd.crosstab(df['embarked'], df['alive'])
alive_class
```

```python
sns.heatmap(alive_class, annot=True, fmt='d', cmap='Blues')
```
